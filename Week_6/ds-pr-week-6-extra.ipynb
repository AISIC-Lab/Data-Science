**Chi-square & Cramér’s V**
```
from scipy.stats import chi2_contingency
tbl = pd.crosstab(df['A'], df['B'])
chi2, p, dof, expected = chi2_contingency(tbl)

def cramers_v(x, y):
    tbl = pd.crosstab(x, y)
    chi2, p, dof, expected = chi2_contingency(tbl)
    n = tbl.values.sum()
    r, k = tbl.shape
    return np.sqrt(chi2 / (n * (min(r-1, k-1))))
```
